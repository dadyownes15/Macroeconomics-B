#### Present value
The present value of a future value, is equivalent to the value, that it takes to reproduce the future value. Some might define this as the value of future money, but the value of future money, is only the present value, because if i had present value of the future money, i would be able to get the future money. 

Present value of a payout $P$ in $t$, with rental rate $r$:

$$PV = P/(1+r)^t = P(1+t)^{-t}$$

This is quite untuitive. If i recieve a P in t=0, P=PV. If i recieve P in t=1, the PV is P/(1+r). This also makes sense, because if i started with P/(1+r) in term 0, i would have gotten aditional rate payouts equivalent which would simply just cancel the dividor.

#### Modeling depth and cash flow

```python
def theta_annuity(N, r):
    return (r / (1 + r)) / (1 - (1 + r) ** (-N))
```

```python
def consumption(T,N,R):
    return T*theta_annuity(N,R)
```
